It provides GPU acceleration, dynamic computation graphs and an intuitive interface for deep learning researchers and developers. 

It's origin was from the Torch library, which had powerful capability of performing tensor-based computations with GPU compatibility, but the only problem was that to code in Torch developers needed to know Lua programming language. The researchers at Meta AI saw this and decided to code the Torch framework in python, since the entire machine learning, data science community knows python very well, and thus came this Pytorch library.

### dataset and dataloader classes

why?
we can simply make a training loop to fetch data but there are problems with that.
1. No standard interface for data
2. No easy way to apply transformations
3. Difficult to do shuffling and resampling
4. Chaotic batch management and parallelization

Dataset class - load data from memory
Dataloader class - make batches from that data loaded from memory and apply transformations on it if required

```python
from torch.utils.data import Dataset, DataLoader

class customDataset(Dataset):

    def __init__(self, X,y):
        self.X = torch.tensor(X, dtype=torch.float32,).reshape(-1,1,28,28) #we need to reshape because we're sending a 2d tensor data not a single tensor
        # -1 is placeholder for batch size 
        # 1 is no. of channels, 
        # 28,28 is the shape of image
        self.y = torch.tensor(y, dtype=torch.long)
    
    def __len__(self):
        return self.X.shape[0]
    
    def __getitem__(self, index):
        return self.X[index], self.y[index]
```

```python
train_dataset = customDataset(X_train,y_train)
test_dataset = customDataset(X_test, y_test)
```

```python
batch_size = 32
```

```python
train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=batch_size, shuffle=False)
```

### using nn modules to define cnn definition

torch.nn module is core library that provides classes and functions to help build neural network efficiently <br>

It abstracts the complexity of creating and training neural networks by offering pre-built layers, loss functions, activation functions, optimizers and other utilities, enabling you to focus on designing and experimenting with model architectures.

Convolutional Neural Network (CNN)
It's one type of neural network that works best on grid based data like images, which is 2d array of pixel values. It can detect spatial patterns and textures seamlessly and that's why we use cnn in every image recognition task, computer vision task, etc.

The model has two stages, first one being the feature extraction stage, where we perform two operations, convolution and pooling, so this extraction stage is made of pairs of convolution and pooling layers, in convolution layers, there are small filters which you apply on the images (those are called convolution operations) to extract low-level features like edges, there can be multiple filters, different filters detect different edges patterns etc.,

Note : Detection in edges is basically change in intensity.

After the convolution step, we get a non-linear feature map, which is then passed to a pooling layer. Pooling layer is used in CNNs to reduce the spatial dimensions (width and height) of the input feature maps while retaining the most important information.

[Demo of Pooling](https://deeplizard.com/resource/pavq7noze3)

Why are Pooling Layers Important? 
1. Dimensionality Reduction: Pooling layers reduce the spatial size of the feature maps, which decreases the number of parameters and computations in the network.
2. Translation Invariance: Pooling helps the network become invariant to small translations or distortions in the input image. For example, even if an object in an image is slightly shifted, the pooled output will remain relatively unchanged. <br>

```python
import torch.nn as nn

class MyNN(nn.Module):
    def __init__(self, input_features):
        
        super().__init__()  # calls nn.Module.__init__() i.e parent class constructor
                            # Initializes nn.Module internals
                            # Registers layers & parameters
                            # Enables .parameters(), .to(device), .train()
        

        self.features = nn.Sequential(
            nn.Conv2d(input_features, 32, kernel_size=3, padding='same'),
            nn.ReLU(), # ReLU adds non-linearity
            nn.BatchNorm2d(64), # Normalizes each channel using batch mean & variance, 
            nn.MaxPool2d(kernel_size=2, stride=2),

            nn.Conv2d(32, 64, kernel_size=3, padding='same'),
            nn.ReLU(),
            nn.BatchNorm2d(64),
            nn.MaxPool2d(kernel_size=2, stride=2)
        )
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(64*7*7, 128),
            nn.ReLU(),
            nn.Dropout(p=0.4), # randomly turn of neurons during forward pass, p = probability

            nn.Linear(128, 64),
            nn.ReLU(),
            nn.Dropout(p=0.4),

            nn.Linear(64, 10)
        )

    def forward(self, x):
        x = self.features(x)
        x = self.classifier(x)

        return x
```

```python
learning_rate = 0.01
epochs = 100
```

```python
import torch.optim as optim

# creating an instance of the model
model = MyNN(1) # 1 as in 1 chanel image

# shifting to model on gpu
model.to(device)

# initializing a loss function here
loss_function = nn.CrossEntropyLoss()

# initializing an optimizer
optimizer = optim.SGD(model.parameters(), lr=learning_rate)
```

Optimization of model hyperparameters
1. No. of hidden layers
2. Neurons per layer
3. Optimizer
5. Learning rate
6. Batch size
7. Dropout rate

### training loop

```python
# training loop

for epoch in range(epochs):

  total_epoch_loss = 0
  
  for batch_features, batch_labels in train_loader:
      # move data to gpu
      batch_features, batch_labels = batch_features.to(device), batch_labels.to(device)

      # forward pass
      outputs = model(batch_features)

      # calculate loss
      loss = loss_function(outputs, batch_labels)

      #setting the accumulated gradients to zero
      optimizer.zero_grad()

      # backward pass
      loss.backward()

      # update grads
      optimizer.step()

      total_epoch_loss = total_epoch_loss + loss.item()

  avg_loss = total_epoch_loss/len(train_loader)
  print(f'Epoch: {epoch + 1}, Loss: {avg_loss}')
```

### evaluating on test data

```python
model.eval()
```

```python
# evaluation on test data
total = 0
correct = 0

with torch.no_grad():

  for batch_features, batch_labels in test_loader:

    # move data to gpu
    batch_features, batch_labels = batch_features.to(device), batch_labels.to(device)

    outputs = model(batch_features)

    _, predicted = torch.max(outputs, 1)
    # torch.max(outputs, 1) returns two things:
    # 1. max value in each row (confidence)
    # 2. index of max value (class label)
    # predicted - class index with highest score for each sample
    # and '_' means don’t care about the max values

    total = total + batch_labels.shape[0]

    correct = correct + (predicted == batch_labels).sum().item()

print(f"Accuracy on test data {correct/total * 100}%")
```

```python
# Get a random image
img, label = test[0] # Grab first image

model.eval()
with torch.inference_mode():
    # Add an extra dimension for batch size [1, 28, 28] -> [1, 1, 28, 28]
    img_input = img.unsqueeze(dim=0).to(device)
    
    # Get prediction logits
    pred_logits = model(img_input)
    
    # Convert logits to prediction labels
    pred_label = torch.argmax(torch.softmax(pred_logits, dim=1), dim=1)

plt.imshow(img.squeeze(), cmap="gray")
plt.title(f"Truth: {label} | Pred: {pred_label.item()}")
plt.axis(False);
```
